## Detailed Architectural Deliverables

### Deliverable 1: Data You Would Need
To build a robust, production-grade text classifier, we need a diverse set of real-world datasets:
1. **Labeled Historical Tickets:** High-quality raw text paired with manually verified tags (e.g., Billing, Tech Issue, General).
2. **Metadata Context:** Timestamps, ticket origin (e.g., web portal, mobile app, API, direct email), user tier status (e.g., enterprise, paying customer, free tier), and conversation depth.
3. **Multilingual and Varied Forms:** Multi-sentence, fragmented expressions, jargon, typos, and negative/positive sentiments to capture customer expression variability.

### Deliverable 2: Preprocessing Steps
*   **Text Cleaning & Normalization:** Conversion of all text to lowercase, stripping out raw HTML tags, URLs, and identifying/anonymizing personally identifiable information (PII) like phone numbers, passwords, and custom tokens.
*   **Tokenization & Stop-word Filtering:** Parsing text into tokens and filtering out words with low informational value to improve processing latency.
*   **Lemmatization/Stemming:** Normalizing varied inflected forms of words (e.g., "crashed", "crashing", "crashes" -> "crash") so features capture underlying concepts rather than specific syntax.
*   **Vectorization Scheme:** Transform textual segments into numeric vectors. While classical approaches like TF-IDF or Bag of Words represent simple statistical baselines, modern approaches leverage dense Vector Embeddings (from systems like OpenAI's `text-embedding-3-small` or Hugging Face's SentenceTransformers) to retain contextual semantic properties.

````python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score
import pandas as pd

# Reload the updated database to ensure new samples are captured
df = pd.read_json("customer_support_db.json")

X = df['message']
y = df['category']

# Stratified split to preserve class proportions on the expanded dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Updated dataset loaded successfully.")
print(f"Training samples: {len(X_train)} | Test samples: {len(X_test)}")
````

### Deliverable 3: Model Choice & Deep Rationale
*   **Baseline Model: TF-IDF + Logistic Regression (Implemented here):** Highly efficient, instantly trainable, and remarkably explainable. It determines class mappings using explicit word coefficients, establishing an ideal standard metric.
*   **Advanced Production Choice: Fine-tuned DistilBERT or LLM APIs:**
    *   *Why BERT:* Transformer-based language architectures leverage self-attention layers to identify key contexts (e.g., recognizing that "The balance won't load" indicates a technical issue rather than a billing request).
    *   *Why LLMs:* Extremely robust out-of-the-box performance using strategic Few-Shot Prompting, avoiding complex model deployment and infrastructure overhead.

````python
# Re-instantiate and train pipeline with English stop words and bigrams
classifier_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english', ngram_range=(1, 2))),
    ('clf', LogisticRegression(random_state=42))
])

# Train model on our expanded dataset
classifier_pipeline.fit(X_train, y_train)

# Perform test evaluations
y_pred = classifier_pipeline.predict(X_test)

print("=== Updated Evaluation Report ===\n")
print(classification_report(y_test, y_pred, zero_division=0))
print(f"Overall Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%")
````

### Deliverable 4: How Actual Solution Will Look (Production Blueprint)
1. **API Gateway:** Microservices collect customer messages from front-end contact channels (web dashboard, chat widgets, Zendesk API) and route them to our model prediction endpoint.
2. **Inference Pipeline:** A serverless function (AWS Lambda, GCP Cloud Run) cleans inputs, tokenizes texts, generates vectors, and invokes the trained classifier model.
3. **Smart Routing Broker:** Categorized classifications trigger immediate actions:
    * *Billing tickets* directly integrate with Stripe/Zuora diagnostics and route to billing teams.
    * *Technical bugs* trigger system performance analysis and route to DevOps/Engineering squads.
    * *General inquiries* draft automated helpful resources to speed up response times.

### Deliverable 5: Evaluation Approach
*   **Offline Validation:** Tracking Macro F1-Score (to address severe class imbalances) alongside Precision (minimizing mistranslating sensitive billing/tech bugs) and Recall.
*   **A/B Testing / Shadow Deployments:** Serving live traffic silently to verify system performance before complete cutover.
*   **Continuous Analytics:** Monitoring manual overrides from support agents to flags instances of incorrect classifications.

### Deliverable 6: Potential Risks & Mitigation
*   **Concept Drift:** New products and features lead to new customer issues. *Mitigation:* Conduct scheduled automated pipeline retraining using new labeled dataset imports.
*   **Extreme Class Imbalance:** Highly common issues can dominate data, reducing prediction rates for minority classes. *Mitigation:* Apply loss-reweighting parameters or synthetic data generation (SMOTE) to normalize predictions.
*   **Sarcasm & Complex Syntax:** Sarcastic complaints can mislead classifiers. *Mitigation:* Integrate sentiment analysis checks as secondary input layers to filter complex tickets.
